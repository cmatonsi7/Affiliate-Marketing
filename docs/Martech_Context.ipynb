# MarTech Domain Context

*Why this project reflects real affiliate marketing / partner
performance economics, not just generic revenue analysis.*

## The business model this data represents

Affiliate/partner marketing programs pay partners (content sites,
influencers, coupon sites, loyalty/cashback platforms) a commission for
traffic or sales they drive — the `commission_paid` field in this dataset.
Unlike a normal sales channel, the program only pays for results, but that
also means partner behavior is highly sensitive to two things this analysis
directly measures:

- **Commission rate competitiveness** — if a partner's effective rate
  (commission ÷ revenue) falls behind what competing programs offer, their
  traffic quietly migrates elsewhere. This shows up as a *revenue decline
  with no obvious external cause* — exactly the pattern the health
  segmentation query surfaces.
- **Partner tier / type dynamics** — different partner types monetize
  differently. Content Sites (SEO-driven blogs/reviews) tend to have large,
  volatile revenue swings tied to search ranking changes and seasonal
  content relevance. Influencers tend to show steadier, more linear growth
  or decline tied to their own audience growth. Coupon/Loyalty sites
  typically show the most stable, incremental patterns since their traffic
  is intent-driven rather than discovery-driven. The forecasting extension
  (`forecasting/`) found exactly this: moving average won for the volatile
  categories (Content Site), linear regression won for the steadier ones
  (Coupon Site, Loyalty/Cashback, Influencer) — a pattern that lines up
  with how these partner types actually behave in a real program, not a
  coincidence of the math.

## Why "At Risk" partners matter disproportionately

In affiliate programs, **revenue concentration risk** is a known,
named problem: a small number of partners typically drive a
disproportionate share of program revenue (see the "Revenue Concentration
Risk" query in `sql/02_business_queries.sql`). When one of those large
partners starts declining, it isn't just "a partner having a bad quarter"
— it can materially move the whole program's numbers. This is why the
health segmentation isn't just a nice-to-have visualization: identifying
which declining partners are large enough to matter, versus which are
small and not worth retention spend, is a genuine prioritization decision
a partnerships/affiliate manager has to make every month.

## Retention economics — what the numbers imply

A program typically has a limited retention/incentive budget (bonus
commission tiers, dedicated account management time, co-marketing spend).
The combination of the health segmentation and revenue concentration
queries answers the real underlying business question: **"if I only have
budget to actively retain 3-5 partners this quarter, which ones give the
best return on that effort?"** — not simply "which partners are declining"
(too broad) or "which partners are biggest" (ignores trend), but the
intersection of both, which is what `sql/02_business_queries.sql` query 1
and 2 together are built to answer.
