## Handling policy

Quarantined rows are never deleted. Each is written to
`analytics.performance_quarantine` with:
- `source_row_id` — traceable back to the original CSV row
- `quarantine_reason` — human-readable, one row can have multiple reasons
- `raw_data` — the full original row as JSONB, for manual review or reprocessing

This mirrors the quarantine pattern used in the production-grade
e-commerce pipeline project, applied here to a relational warehouse instead
of a file-based pipeline.

## Recommendation

The two dominant failure modes (missing `commission_paid`, missing
`revenue`) suggest an upstream issue in whatever system exports this data —
worth flagging to the data source owner rather than treating as one-off
noise, since it recurs at a similar rate across the dataset.
